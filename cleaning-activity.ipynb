```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
```

## 01 Load data

```python
raw = pd.read_excel('usedcars_noisy.xlsx')
print(raw.shape)

# --- output: 160 rows, 6 columns
```

## 02 Display contents and find noisy rows

```python
raw.head(10) # display first 10 rows of the dataset
```

### Identifying noisy rows
Using boolean mask to identify noisy rows. Each instance will be flagged if true, and false if no noise is found

```python
# identifying noisy rows in price and mileage columns
noisy = (raw["price"] == 0) | (raw["mileage"] == -1)

noisy # displaying the boolean series of noisy rows
```

```python
# printing the number of noisy rows in price and mileage columns, including total
print("price = 0:", (raw["price"] == 0).sum(), "rows")
print("mileage = -1:", (raw["mileage"] == -1).sum(), "rows")
print("total noisy rows:", noisy.sum(), "rows") # only sums true values

raw[noisy] # displaying all the noisy rows

# --- output: 2 rows contain noise values in both price and mileage, so there are 10 rows in total
```

## 03 Replacing noisy values

```python
mean_price = raw["price"].mean() # calculating the mean of price column
median_mileage = raw["mileage"].median() # calculating the median of mileage column
print("mean price:", mean_price)
print("median mileage:", median_mileage)

# replacing noisy values in price and mileage columns with mean and median respectively
raw["price"] = raw["price"].replace(0, mean_price)
raw["mileage"] = raw["mileage"].replace(-1, median_mileage) 

# printing the number of noisy rows in price and mileage columns after replacement
print("price = 0:", (raw["price"] == 0).sum(), "rows")
print("mileage = -1:", (raw["mileage"] == -1).sum(), "rows") 
raw[noisy] # displaying all the noisy rows after replacement

# --- output: all noisy rows have been replaced with mean and median values
```

## 04 Save cleaned data

```python
raw.to_csv("usedcars_cleaned.csv", index=False) # saving the cleaned dataset to a new CSV file
print("saved", len(raw), "rows to usedcars_cleaned.csv")

cleaned_data = raw.copy() # creating a copy of the cleaned dataset
```

## 06 Histograms

```python
fig, axes = plt.subplots(1, 3, figsize=(11, 3.2)) # creating a figure with 1 row and 3 columns of subplots, with a specified figure size

for ax, col in zip(axes, ["price", "mileage", "year"]): # ax is the current axis, col is the current column name
    bins = range(2000, 2014) if col == "year" else 20 # setting the number of bins for the histogram based on the column
    ax.hist(cleaned_data[col], bins=bins, color="red" if col == "year" else "orange", edgecolor="white", linewidth=2)
    ax.set_title(col.capitalize(), fontsize=10)
    ax.set_ylabel("Number of cars", fontsize=8)
    ax.tick_params(labelsize=8)
    ax.spines[["top", "right"]].set_visible(False) # hides top and right spines of the plot

fig.suptitle("Distribution of Price, Mileage, and Year", fontsize=12)
plt.tight_layout()
plt.savefig("usedcars_cleaned_histograms.png", dpi=150)
plt.show()

#  --- output analysis: 
```
